```python
order_details=pd.read_csv(filepath_or_buffer="data.tsv.txt",sep="\t")
order_details
```

```python
order_details.groupby(by="item_name")["quantity"].sum().sort_values(ascending= True).head()
```

```python
order_details.groupby(by="item_name")["quantity"].sum().sort_values(ascending= False).head()
```

```python
order_details["item_price"] = (
    order_details["item_price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.split()
    .str[0]
    .astype(float)
)
order_details.dtypes
```

```python
order_details["revenue"] = (
    order_details["quantity"] * order_details["item_price"]
)
```

```python
item_revenue = (
    order_details
    .groupby("item_name")
    .agg(
        total_quantity=("quantity", "sum"),
        individual_revenue=("revenue", "sum")
    )
    .reset_index()
)
```

```python
top_5_items = (
    item_revenue
    .sort_values(by="individual_revenue", ascending=False)
    .head(5)
)

print(top_5_items)
```

```python
total_revenue_top_5 = top_5_items["individual_revenue"].sum()
print("Total revenue of top 5 items:", total_revenue_top_5)
```
